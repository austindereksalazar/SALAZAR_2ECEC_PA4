A. VISAYAS COMMUNICATION DATAFRAME

```python
import pandas as pd
import matplotlib.pyplot as plt

df_raw = pd.read_excel('board2.xlsx')
df = df_raw.copy()
df['Average'] = df[['Math', 'GEAS', 'Electronics', 'Communication']].mean(axis=1)
```

```python

# Create VisComm DataFrame with explicit conditions
VisComm = df[(df['Hometown'] == 'Visayas') & (df['Track'] == 'Communication')]

# Select specific columns in the exact requested order
VisComm = VisComm[['Name', 'Gender', 'Math', 'Electronics', 'Average']]

# Display DataFrame and its number of rows
display(VisComm)
print(f"Number of rows in VisComm DataFrame: {len(VisComm)}")
```

B. VISAYAS FEMALE DATAFRAME

```python
# Create VisFemale DataFrame with explicit conditions
VisFemale = df[(df['Hometown'] == 'Visayas') & (df['Gender'] == 'Female')]

# Select specific columns
VisFemale = VisFemale[['Name', 'Track', 'GEAS', 'Electronics', 'Average']]

# Display VisFemale DataFrame
print("VisFemale DataFrame:")
display(VisFemale)

# Display rows of VisFemale whose Average is at least 60 (without overwriting)
print("\nVisFemale students with an Average of at least 60:")
display(VisFemale[VisFemale['Average'] >= 60])
```

C. CATEGORY-AVERAGE VISUALIZATION

```python
# a. Compute the mean of Average for every category
track_mean = df.groupby('Track')['Average'].mean().reset_index()
gender_mean = df.groupby('Gender')['Average'].mean().reset_index()
hometown_mean = df.groupby('Hometown')['Average'].mean().reset_index()

# b. Display the three summary tables
print("Mean Average by Track:")
display(track_mean)

print("\nMean Average by Gender:")
display(gender_mean)

print("\nMean Average by Hometown:")
display(hometown_mean)

# c. Create one figure containing three bar charts
fig, axes = plt.subplots(1, 3, figsize=(15, 5), sharey=True) # ensures a consistent vertical scale appropriate for comparing all three plots

# 1. Bar chart for Track
axes[0].bar(track_mean['Track'], track_mean['Average'], color='skyblue')
axes[0].set_title('Mean Average by Track')
axes[0].set_xlabel('Track')
axes[0].set_ylabel('Mean Average')
axes[0].tick_params(axis='x', rotation=45) # Readable category labels

# 2. Bar chart for Gender
axes[1].bar(gender_mean['Gender'], gender_mean['Average'], color='lightgreen')
axes[1].set_title('Mean Average by Gender')
axes[1].set_xlabel('Gender')

# 3. Bar chart for Hometown
axes[2].bar(hometown_mean['Hometown'], hometown_mean['Average'], color='salmon')
axes[2].set_title('Mean Average by Hometown')
axes[2].set_xlabel('Hometown')

# Adjust layout so labels don't overlap
plt.tight_layout()
plt.show()
```

Interpretations:
* Track: Based on the observed dataset, the Communication track has the highest sample mean Average among the academic tracks.
* Gender: Based on the observed dataset, Male students have a higher sample mean Average compared to the other group.
* Hometown: Based on the observed dataset, students from Luzon have the highest sample mean Average compared to other hometowns.
